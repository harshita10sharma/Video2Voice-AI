```python
import os

REFERENCE = "laksh_reference.wav"

print("Reference exists:", os.path.exists(REFERENCE))

if os.path.exists(REFERENCE):
    print("Reference size:", os.path.getsize(REFERENCE), "bytes")
```

```python
import torch
from chatterbox.tts import ChatterboxTTS

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading Chatterbox...")
print("Device:", DEVICE)

model = ChatterboxTTS.from_pretrained(device=DEVICE)

print("================================")
print("CHATTERBOX MODEL LOADED")
print("================================")
```

```python
import torchaudio as ta

test_text = """
Hello everyone. Today I want to share a very simple story with you.
Sometimes the smallest moments in our lives can teach us the biggest lessons.
"""

print("Generating voice test...")

wav = model.generate(
    test_text,
    audio_prompt_path=REFERENCE,
    exaggeration=0.5,
    cfg_weight=0.5,
    temperature=0.8,
)

ta.save(
    "laksh_voice_test.wav",
    wav,
    model.sr
)

print("Voice test saved.")
print("Sample rate:", model.sr)
```
