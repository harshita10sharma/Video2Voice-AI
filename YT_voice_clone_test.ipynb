```python
import torch
from chatterbox.tts import ChatterboxTTS

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading Chatterbox model...")
print("Device:", device)

model = ChatterboxTTS.from_pretrained(device=device)

print("Chatterbox model loaded successfully!")
```

```python
import os

print("Reference file exists:", os.path.exists("laksh_reference.wav"))

if os.path.exists("laksh_reference.wav"):
    print("Reference file size:",
          round(os.path.getsize("laksh_reference.wav") / (1024 * 1024), 2),
          "MB")
```

```python
import torchaudio

test_text = (
    "Hello everyone. This is a test of the English voice "
    "generated for our automated video dubbing system."
)

print("Generating test speech...")

wav = model.generate(
    test_text,
    audio_prompt_path="laksh_reference.wav"
)

output_file = "chatterbox_test.wav"

torchaudio.save(
    output_file,
    wav,
    model.sr
)

print("Test audio saved:", output_file)
```
