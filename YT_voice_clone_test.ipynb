```python
import torch
from chatterbox.tts import ChatterboxTTS

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading Chatterbox model...")
print("Device:", device)

model = ChatterboxTTS.from_pretrained(device=device)

print("Chatterbox model loaded successfully!")
```

```python
import torchaudio

test_text = (
    "Hello everyone. This is a test of the English voice "
    "generated for our automated video dubbing system."
)

print("Generating test speech...")

wav = model.generate(
    test_text,
    audio_prompt_path="laksh_reference.wav"
)

output_file = "chatterbox_test.wav"

torchaudio.save(
    output_file,
    wav,
    model.sr
)

print("Test audio saved:", output_file)
```

```python
import json
import os
import torchaudio

with open("translated.json", "r", encoding="utf-8") as f:
    translated_segments = json.load(f)

os.makedirs("tts_segments", exist_ok=True)

print("Generating first 3 translated segments...\n")

for i, segment in enumerate(translated_segments[:3]):

    text = segment.get("translated", "").strip()

    if not text:
        print(f"Segment {i}: EMPTY - skipped")
        continue

    print(
        f"Segment {i}: "
        f"{segment['start']}s -> {segment['end']}s"
    )

    print("Text:", text)

    wav = model.generate(
        text,
        audio_prompt_path="laksh_reference.wav"
    )

    output_file = f"tts_segments/segment_{i:04d}.wav"

    torchaudio.save(
        output_file,
        wav.cpu(),
        model.sr
    )

    print("Saved:", output_file)
    print()
```

```python
import os
import json
import torch
import torchaudio

# Load translated segments
with open("translated.json", "r", encoding="utf-8") as f:
    segments = json.load(f)

os.makedirs("tts_segments", exist_ok=True)

print("Total segments:", len(segments))
print("Starting TTS generation...\n")

for i, segment in enumerate(segments):

    text = segment["translated"].strip()

    if not text:
        print(f"[{i+1}/90] Empty segment - skipped")
        continue

    output_file = f"tts_segments/segment_{i:04d}.wav"

    # Skip if already generated
    if os.path.exists(output_file):
        print(f"[{i+1}/90] Already exists: {output_file}")
        continue

    print(f"[{i+1}/90]")
    print(f"Text: {text}")

    wav = model.generate(
        text=text,
        audio_prompt_path="laksh_reference.wav"
    )

    torchaudio.save(
        output_file,
        wav.cpu(),
        model.sr
    )

    print(f"Saved: {output_file}\n")

print("====================================")
print("ALL TTS SEGMENTS GENERATED")
print("====================================")
```
